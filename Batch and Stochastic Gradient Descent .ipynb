### Problem: Predict y using y = mx + c

```python
# Sample training data
X = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]  # y = 2x
```

#### Batch Gradient Descent for Linear Regression

```python
# Batch Gradient Descent
m = 0
c = 0
L = 0.01
epochs = 1000
n = len(X)

for _ in range(epochs):
    dm = 0
    dc = 0
    for i in range(n):
        y_pred = m * X[i] + c
        error = y_pred - y[i]
        dm += error * X[i]
        dc += error
    m -= L * (dm / n)
    c -= L * (dc / n)

# Save result
m_b = m
c_b = c
```

#### Stochastic Gradient Descent for Linear Regression

```python
# Stochastic Gradient Descent
m = 0
c = 0

for _ in range(epochs):
    for i in range(len(X)):
        y_pred = m * X[i] + c
        error = y_pred - y[i]
        m -= L * error * X[i]
        c -= L * error

# Save result
m_s = m
c_s = c
```

#### Plot

```python
import matplotlib.pyplot as plt

y_pred_batch = [m_b * xi + c_b for xi in X]
y_pred_stochastic = [m_s * xi + c_s for xi in X]

plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X, y_pred_batch, color='green', label='Batch Gradient Descent')
plt.plot(X, y_pred_stochastic, color='red', linestyle='--', label='Stochastic Gradient Descent')

plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear Regression using GD Methods")
plt.legend()
plt.grid(True)
plt.show()
```
